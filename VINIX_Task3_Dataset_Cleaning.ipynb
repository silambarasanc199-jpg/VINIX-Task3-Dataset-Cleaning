```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import zipfile
import requests
import io

sns.set_theme(style="whitegrid")
```

```python
url = "https://archive.ics.uci.edu/static/public/320/student+performance.zip"

response = requests.get(url)
with zipfile.ZipFile(io.BytesIO(response.content)) as outer_zip:
    # Extract the nested 'student.zip' from the main ZIP archive
    with outer_zip.open("student.zip") as nested_zip_bytes:
        with zipfile.ZipFile(io.BytesIO(nested_zip_bytes.read())) as inner_zip:
            # Read 'student-mat.csv' from the nested archive
            with inner_zip.open("student-mat.csv") as file:
                df = pd.read_csv(file, sep=";")

print("Original Dataset Shape:", df.shape)
display(df.head())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()
```

```python
df_clean = df.copy()

# Remove duplicate rows
df_clean = df_clean.drop_duplicates()

# Fill missing numeric values with median
for col in df_clean.select_dtypes(include=np.number).columns:
    df_clean[col] = df_clean[col].fillna(df_clean[col].median())

# Fill missing categorical values with mode
for col in df_clean.select_dtypes(include="object").columns:
    if df_clean[col].isnull().any():
        df_clean[col] = df_clean[col].fillna(df_clean[col].mode()[0])

print("Cleaning completed!")
print("Cleaned Dataset Shape:", df_clean.shape)
print("Remaining Missing Values:", df_clean.isnull().sum().sum())
print("Remaining Duplicates:", df_clean.duplicated().sum())
```

```python
comparison = pd.DataFrame({
    "Metric": ["Rows", "Columns", "Missing Values", "Duplicate Rows"],
    "Before Cleaning": [
        df.shape[0],
        df.shape[1],
        df.isnull().sum().sum(),
        df.duplicated().sum()
    ],
    "After Cleaning": [
        df_clean.shape[0],
        df_clean.shape[1],
        df_clean.isnull().sum().sum(),
        df_clean.duplicated().sum()
    ]
})

display(comparison)
```

```python
df_clean.to_csv("student_performance_cleaned.csv", index=False)

print("Cleaned dataset saved successfully!")
```
